```python
## Carga de datos: importa librerias, lee el TSV y muestra primeras filas
## Ajusta 'data_path' si cambias la ubicacion del archivo; 'na_vals' captura nulos comunes.
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

try:
    import seaborn as sns
    sns.set_theme(context='notebook', style='whitegrid')
    HAS_SEABORN = True
except Exception:
    HAS_SEABORN = False

pd.set_option('display.max_columns', 100)
pd.set_option('display.width', 120)

data_path = 'pheno_data/input_raw_data_winter.tsv'
na_vals = ['', 'NA', 'NaN', 'na', 'N/A', 'null', 'NULL']
df = pd.read_csv(data_path, sep='\t', na_values=na_vals)
df.sample(25)
```

```python
## Estructura y tipos: dimensiones, lista de columnas y dtypes
## Ayuda a validar schema y nombres antes de analizar
print(f'Filas: {df.shape[0]}  |  Columnas: {df.shape[1]}')

print('\nColumnas:')
for i, c in enumerate(df.columns, 1):
    print(f'{i:2d}. {c}')

print('\nTipos de datos:')
df.dtypes
```

```python
## Valores perdidos: conteo de nulos por columna (ordenado)
## Sirve para decidir imputacion o filtrado por rasgo
missing = df.isna().sum().sort_values(ascending=False)
missing[missing > 0].to_frame(name='num_missing')
```

```python
## Descriptivos: columnas numericas/categoricas y resumen estadistico
## Revisa rangos, medias y posibles valores anómalos
num_cols = df.select_dtypes(include=[np.number]).columns
cat_cols = df.select_dtypes(exclude=[np.number]).columns

print('Columnas numericas:', list(num_cols))
print('Columnas categoricas:', list(cat_cols))

print('\nResumen numerico:')
df[num_cols].describe().T
```

```python
## Categoricas: resumen y conteo de valores en trial/status
## Permite ver distribucion de grupos y categorias ausentes
print('Resumen categorico:')
if len(cat_cols) > 0:
    display(df[cat_cols].describe().T)
else:
    print('No hay columnas categoricas.')

for col in ['trial', 'status']:
    if col in df.columns:
        print(f"\nDistribucion de valores para '{col}':")
        display(df[col].value_counts(dropna=False).to_frame(name='count'))
```

```python
## Correlaciones: calcula matriz y dibuja heatmap
## Si ejecutas esta celda sola, se recalculan columnas numericas
if 'num_cols' not in locals():
    num_cols = df.select_dtypes(include=[np.number]).columns
corr = df[num_cols].corr(method='pearson')
display(corr)

plt.figure(figsize=(10, 8))
if HAS_SEABORN:
    sns.heatmap(corr, cmap='coolwarm', annot=True, square=True, cbar_kws={'shrink': 0.8})
else:
    plt.imshow(corr, cmap='coolwarm', interpolation='nearest')
    plt.colorbar(shrink=0.8)
    plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
    plt.yticks(range(len(corr.index)), corr.index)
plt.title('Matriz de correlacion (features numericas)')
plt.tight_layout()
plt.show()
```

```python
## Top correlaciones: pares unicos con mayor |corr|
## Útil para detectar redundancia o relaciones fuertes entre rasgos
if 'corr' not in locals():
    corr = df.select_dtypes(include=[np.number]).corr()
mask = np.triu(np.ones(corr.shape), k=1).astype(bool)
top_pairs = corr.abs().where(mask).stack().sort_values(ascending=False)
top_pairs.head(15).to_frame(name='|corr|')
```

```python
## Distribuciones por grupo (trial, status)
## Hist/KDE y violin/box por trait y grupo
import numpy as np
import matplotlib.pyplot as plt
try:
    import seaborn as sns
    _HAS_SNS = True
    sns.set_style("whitegrid")
except Exception:
    _HAS_SNS = False

def _safe_palette(n):
    if _HAS_SNS:
        return sns.color_palette(n_colors=n)
    base = ["#1f77b4","#ff7f0e","#2ca02c","#d62728","#9467bd","#8c564b","#e377c2","#7f7f7f","#bcbd22","#17becf"]
    if n <= len(base):
        return base[:n]
    reps = int(np.ceil(n / len(base)))
    return (base * reps)[:n]

def plot_distributions_all_traits(df, trait_cols=None, groups=("trial","status"), bins=30, kde=True, figsize=(12,5), dpi=110, show=True, outdir=None):
    # Detectar traits numéricos si no se pasan
    if trait_cols is None:
        trait_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    trait_cols = [c for c in trait_cols if df[c].notna().sum() > 1 and df[c].var(skipna=True) > 0]
    saved = {}
    import os
    if outdir is not None:
        os.makedirs(outdir, exist_ok=True)
    for trait in trait_cols:
        for group_col in groups:
            if group_col not in df.columns:
                continue
            sub = df[[group_col, trait]].dropna()
            if sub.empty:
                continue
            cats = sub[group_col].astype(str).unique().tolist()
            if len(cats) < 2:
                continue
            fig, axes = plt.subplots(1, 2, figsize=figsize)
            ax0 = axes[0]
            pal = _safe_palette(len(cats))
            if _HAS_SNS:
                for i, cat in enumerate(cats):
                    dcat = sub[sub[group_col].astype(str) == cat][trait]
                    sns.histplot(dcat, ax=ax0, bins=bins, stat="density", kde=kde, color=pal[i], label=str(cat), alpha=0.45)
                ax0.legend(title=group_col)
            else:
                for i, cat in enumerate(cats):
                    dcat = sub[sub[group_col].astype(str) == cat][trait]
                    ax0.hist(dcat, bins=bins, density=True, alpha=0.45, color=pal[i], label=str(cat))
                ax0.legend(title=group_col)
            ax0.set_title(f"{trait} — Hist/KDE por {group_col}")
            ax0.set_xlabel(trait)
            ax0.set_ylabel("Densidad")
            ax1 = axes[1]
            if _HAS_SNS:
                sns.violinplot(data=sub, x=group_col, y=trait, ax=ax1, inner="quartile", palette=pal)
            else:
                data_groups = [sub[sub[group_col].astype(str) == cat][trait].values for cat in cats]
                bp = ax1.boxplot(data_groups, labels=cats, patch_artist=True)
                for patch, color in zip(bp["boxes"], pal):
                    patch.set_facecolor(color)
            ax1.set_title(f"{trait} — Violin/Box por {group_col}")
            ax1.set_xlabel(group_col)
            ax1.set_ylabel(trait)
            fig.suptitle(f"Distribuciones de {trait} por {group_col}")
            fig.tight_layout()
            path = None
            if outdir is not None:
                safe = lambda s: "".join(ch if ch.isalnum() or ch in ("-","_") else "_" for ch in s)
                fname = f"dist_{safe(trait)}_by_{safe(group_col)}.png"
                path = os.path.join(outdir, fname)
                fig.savefig(path, dpi=dpi, bbox_inches="tight")
            if show:
                plt.show()
            else:
                plt.close(fig)
            saved[(trait, group_col)] = path
    return saved

# Ejecuta para ver distribuciones por 'trial' y 'status'
plot_distributions_all_traits(df, groups=("trial","status"))
```
